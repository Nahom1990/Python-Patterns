The primary difference between Functional Programming (FP) and Object-Oriented Programming (OOP) lies in how they treat data and state. OOP bundles data and its associated behavior together into objects, whereas FP treats data and functions as completely separate entities, focusing on transforming data through pure mathematical functions. The primary difference between Procedural Programming and the other two paradigms is its focus on a linear sequence of steps. While OOP organizes code around objects and FP organizes it around mathematical functions, Procedural programming organizes code around procedures (or routines) that execute sequentially to complete a task.Here is a direct comparison of all three paradigms:FeatureProcedural ProgrammingObject-Oriented (OOP)Functional (FP)Core ConceptSequence of execution: A step-by-step recipe or checklist.Real-world entities: Bundles of data and behavior.Data transformation: Passing data through a pipeline of functions.Data & LogicSeparated, but functions have direct access to modify global variables.Combined into objects via encapsulation.Separated, and functions never modify data directly (immutability).State ManagementHighly mutable: Values of variables change frequently throughout execution.Mutable: State is contained within objects and modified by methods.Immutable: State never changes; new data states are returned.Primary ToolProcedures, functions, and control loops (for, while).Classes, objects, and methods.Pure functions and recursion.Code ReusabilityLow to moderate (functions can be reused, but they often rely on global state).High (achieved through inheritance and polymorphism).Very high (achieved through small, independent, modular functions).Iconic LanguagesC, Fortran, Pascal, COBOL.Java, C#, C++.Haskell, Clojure, Elixir.How Each Paradigm Solves the Same ProblemImagine you want to calculate and apply a discount to a shopping cart:Procedural: You write a list of instructions. First, look at the cart array. Next, run a for loop to add up the prices. Then, apply a global discount percentage. Finally, overwrite the totalPrice variable with the new number.Object-Oriented: You create a Cart object. It holds its own items and a totalPrice property. You call the method cart.applyDiscount(0.10), which internally changes its own totalPrice.Functional: You pass an immutable cart data structure into a pure function: applyDiscount(cart, 0.10). The function calculates the math and returns an entirely new, updated cart dataset without changing the original one.
