CQRS (Command Query Responsibility Segregation), introduced by Greg Young, is an architectural pattern that separates the code that reads data (Queries) from the code that mutates data (Commands).

In traditional architectures (like RSC, Clean, or Onion), the same domain model and database schema are used for both reads and writes. CQRS breaks this symmetry to allow read and write operations to scale, optimize, and evolve independently.

The Fundamental Shift
TRADITIONAL (SHARED) MODEL:
[ Client ] ──► [ Controller ] ──► [ Unified Service ] ──► [ Single Domain Model ] ──► [ Database ]

CQRS (SEGREGATED) MODEL:
                                 ┌──► [ Command Service ] ──► [ Write Model ] ──► [ Write DB ]
                                 │                                                     │
[ Client ] ──► [ Dispatcher ] ───┤                                             Sync / Events
                                 │                                                     ▼
                                 └──► [ Query Service ]   ──► [ Read Model  ] ──► [ Read DB ]
Commands (State Mutations):

Express intent in the imperative mood (CreateOrder, UpgradeUserPlan, CancelSubscription).

Perform validation and execute business rules.

Do not return data (or return only a minimal status/ID).

Queries (Data Retrieval):

Express data requests (GetOrderDetails, ListActiveUsers).

Bypasses business validation and domain models entirely.

Never mutate state. Returns read-optimized DTOs directly.

CQRS Implementation Levels
Level 1: Single Database / Schema Separation
Commands and Queries are separated in code (handlers/services), but both read and write against the same physical database table.

Level 2: Database / Storage Separation (Eventual Consistency)
Write Database: Optimized for transactional integrity and normalization (e.g., PostgreSQL in 3NF).

Read Database: Optimized for high-speed queries and projections (e.g., MongoDB, Elasticsearch, Redis, or denormalized SQL views).

Sync Engine: Changes made to the Write DB are published via Domain Events or CDC (Change Data Capture) to update the Read DB asynchronously (Eventual Consistency).

When to Use (and Avoid) CQRS
Use CQRS When:
    High Read/Write Asymmetry: Your application has a 100:1 read-to-write ratio (e.g., e-commerce catalogs, news feeds).

    Complex Domain Queries: Read queries require joining 10+ tables or complex aggregations that slow down write models.

    Independent Scaling: You need to scale read servers horizontally (e.g., Read Replicas, Redis caches) while keeping write servers tightly constrained for transactional consistency.

    Combined with Event Sourcing: CQRS is the standard mechanism to query state when writes are stored as a sequence of historical immutable events.

Avoid CQRS When:
    Standard CRUD Applications: If your read and write models look almost identical, CQRS adds massive unnecessary complexity.

    Strict Real-Time Consistency Required: If your users cannot tolerate even sub-second delays between submitting a write and seeing the read updated (Eventual Consistency window).
