Onion Architecture (introduced by Jeffrey Palermo in 2008) is another concentric architectural pattern built on the Dependency Inversion Principle.

Like Hexagonal and Clean Architecture, it rejects the traditional top-down database-centric model (where the database sits at the bottom and everything depends on it) and instead puts the Domain Model at the very center, with all external concerns pointing inward.
