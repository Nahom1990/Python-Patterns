Onion Architecture (introduced by Jeffrey Palermo in 2008) is another concentric architectural pattern built on the Dependency Inversion Principle.

Like Hexagonal and Clean Architecture, it rejects the traditional top-down database-centric model (where the database sits at the bottom and everything depends on it) and instead puts the Domain Model at the very center, with all external concerns pointing inward.

The Onion Layers

│  4. Infrastructure & UI (Web APIs, DB, ORM, Mail, Queue)     │
│    ┌──────────────────────────────────────────────────────┐  │
│    │  3. Application Services / Use Cases                 │  │
│    │    ┌──────────────────────────────────────────────┐  │  │
│    │    │  2. Domain Services (Interfaces)             │  │  │
│    │    │    ┌────────────────────────────────────┐    │  │  │
│    │    │    │  1. Domain Model / Entities        │    │  │  │
│    │    │    └────────────────────────────────────┘    │  │  │
│    │    └──────────────────────────────────────────────┘  │  │
│    └──────────────────────────────────────────────────────┘  │
└──────────────────────────────────────────────────────────────┘
                              │
                    DEPENDENCY DIRECTION
                        INWARD (──►)

1. Domain Model (Core Center)
Contains pure Entities, Value Objects, and business state rules.

Has zero dependencies—no framework code, no ORM annotations, no external library imports.

2. Domain Services
Contains business operations that involve multiple domain entities or core domain contracts (interfaces).

Defines abstract interfaces for repositories and external integrations (e.g., IUserRepository, INotificationService).

3. Application Services
Defines the use cases of the application (e.g., RegisterUserUseCase, CheckoutOrder).

Coordinates domain entities and domain services to accomplish workflow goals.

Depends only on the Domain Services and Domain Model layers.

4. Infrastructure & UI (Outer Ring)
Contains all external concerns: database implementations (SQLAlchemy, EF Core), HTTP controllers, UI components, background job workers, and third-party API clients.

Implements the interfaces defined in the inner layers.
