The Clean Architecture Rings
Clean Architecture organizes an application into four concentric rings:

┌─────────────────────────────────────────────────────────────┐
│ 1. Frameworks & Drivers (Web, DB, UI, External Interfaces) │
│   ┌─────────────────────────────────────────────────────┐   │
│   │ 2. Interface Adapters (Controllers, Gateways)       │   │
│   │   ┌─────────────────────────────────────────────┐   │   │
│   │   │ 3. Application Business Rules (Use Cases)   │   │   │
│   │   │   ┌─────────────────────────────────────┐   │   │   │
│   │   │   │ 4. Enterprise Business Rules        │   │   │   │
│   │   │   │    (Entities)                       │   │   │   │
│   │   │   └─────────────────────────────────────┘   │   │   │
│   │   └─────────────────────────────────────────────┘   │   │
│   └─────────────────────────────────────────────────────┘   │
└─────────────────────────────────────────────────────────────┘
                             │
                  DEPENDENCY DIRECTION
                       INWARD (──►)
Entities (Enterprise Business Rules): Enapsulate high-level, framework-agnostic business rules and data structures. These change least often.

Use Cases (Application Business Rules): Contains application-specific workflows that orchestrate the flow of data to and from Entities to fulfill a user request.

Interface Adapters: Converts data from the format most convenient for Use Cases and Entities into the format required by external agents (e.g., converting HTTP requests into Use Case inputs, or ORM models into Domain Entities).

Frameworks & Drivers: Outer layer consisting of tools, frameworks, databases, and UI devices.

How They Relate Conceptually
       CONCEPTUAL MAPPING OF TERMINOLOGY ACROSS PATTERNS

     Clean Architecture        Hexagonal Architecture         RSC Architecture
  ┌──────────────────────┐    ┌─────────────────────┐    ┌──────────────────────┐
  │ Entities             │    │ Core Domain         │    │ Entities / Domain    │
  │ Use Cases            │ ◄─►│ (Inside)            │ ◄─►│ Service Layer        │
  ├──────────────────────┤    ├─────────────────────┤    ├──────────────────────┤
  │ Interface Adapters   │    │ Ports & Adapters    │    │ Controllers &        │
  │ Frameworks & Drivers │ ◄─►│ (Outside)           │ ◄─►│ Repositories         │
  └──────────────────────┘    └─────────────────────┘    └──────────────────────┘
Hexagonal & Clean Architecture are structural siblings: Clean Architecture is largely a more prescriptive, formalized variant of Hexagonal Architecture. Hexagonal focuses on the Inside/Outside boundary with Ports/Adapters, while Clean explicitly splits the inside into Entities vs. Use Cases and the outside into Adapters vs. Frameworks.

RSC is simpler but vulnerable to tight coupling: In RSC, because the Service sits above the Repository, developers naturally write code where business logic depends directly on database models or ORM methods.

The Core Fix: Both Hexagonal and Clean invert the dependency: the database becomes a plugin to the business logic, rather than the foundation the business logic is built upon.
